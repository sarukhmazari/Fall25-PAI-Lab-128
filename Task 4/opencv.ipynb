```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Load image
image = cv2.imread(r'D:\University\4th semester\PAI Lab\Task 4\dog.jpg')

# ------------------- Operations -------------------

# 1. Erosion
kernel = np.ones((5,5), np.uint8)
erosion = cv2.erode(image, kernel, iterations=1)

# 2. Dilation
dilation = cv2.dilate(image, kernel, iterations=1)

# 3. Bilateral Filtering (smooth while keeping edges)
bilateral = cv2.bilateralFilter(image, d=9, sigmaColor=75, sigmaSpace=75)

# 4. Denoising Colored Image
denoised = cv2.fastNlMeansDenoisingColored(image, None, h=10, hColor=10, templateWindowSize=7, searchWindowSize=21)

# 5. Filter Color (e.g., isolate blue color)
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
# Define blue range
lower_blue = np.array([100, 150, 0])
upper_blue = np.array([140, 255, 255])
mask_blue = cv2.inRange(hsv, lower_blue, upper_blue)
blue_filtered = cv2.bitwise_and(image, image, mask=mask_blue)

# 6. Morphological Operations (Opening and Closing)
opening = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel)
closing = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel)

# ------------------- Display -------------------
plt.figure(figsize=(15,10))

images = [image, erosion, dilation, bilateral, denoised, blue_filtered, opening, closing]
titles = ['Original', 'Erosion', 'Dilation', 'Bilateral Filter', 'Denoised', 'Blue Filtered', 'Opening', 'Closing']

for i in range(len(images)):
    plt.subplot(2,4,i+1)
    plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB))
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Load main image
image = cv2.imread(r'D:\University\4th semester\PAI Lab\Task 4\dog.jpg')

# ------------------- 1. Image Inpainting -------------------
# Create a mask for inpainting (white area = region to inpaint)
mask = np.zeros(image.shape[:2], dtype=np.uint8)
mask[50:150, 50:150] = 255  # Example region
inpainted = cv2.inpaint(image, mask, 3, cv2.INPAINT_TELEA)

# ------------------- 2. Image Registration -------------------
# For demo, shift image to simulate misalignment
rows, cols = image.shape[:2]
M = np.float32([[1, 0, 50], [0, 1, 30]])
shifted = cv2.warpAffine(image, M, (cols, rows))
# Register shifted image back to original using ECC (enhanced correlation coefficient)
gray_ref = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
gray_shifted = cv2.cvtColor(shifted, cv2.COLOR_BGR2GRAY)
warp_matrix = np.eye(2, 3, dtype=np.float32)
criteria = (cv2.TERM_CRITERIA_EPS | cv2.TERM_CRITERIA_COUNT, 5000, 1e-10)
(cc, warp_matrix) = cv2.findTransformECC(gray_ref, gray_shifted, warp_matrix, cv2.MOTION_TRANSLATION)
registered = cv2.warpAffine(shifted, warp_matrix, (cols, rows), flags=cv2.INTER_LINEAR + cv2.WARP_INVERSE_MAP)

# ------------------- 3. Background Subtraction -------------------
fgbg = cv2.createBackgroundSubtractorMOG2()
fg_mask = fgbg.apply(image)

# ------------------- 4. Background Subtraction using Running Average -------------------
avg_bg = np.float32(image)
# Simulate updating background with running average
cv2.accumulateWeighted(image, avg_bg, 0.5)
bg_subtracted = cv2.convertScaleAbs(avg_bg)

# ------------------- 5. Foreground Extraction using GrabCut -------------------
mask_gc = np.zeros(image.shape[:2], np.uint8)
rect = (50, 50, 200, 200)  # Rectangle around object to extract
bgdModel = np.zeros((1,65), np.float64)
fgdModel = np.zeros((1,65), np.float64)
cv2.grabCut(image, mask_gc, rect, bgdModel, fgdModel, 5, cv2.GC_INIT_WITH_RECT)
mask2 = np.where((mask_gc==2)|(mask_gc==0), 0, 1).astype('uint8')
grabcut_result = image * mask2[:, :, np.newaxis]

# ------------------- Display -------------------
plt.figure(figsize=(18,10))
images = [image, inpainted, shifted, registered, fg_mask, bg_subtracted, grabcut_result]
titles = ['Original', 'Inpainted', 'Shifted', 'Registered', 'Background Subtraction (MOG2)', 
          'Running Avg Background', 'GrabCut Foreground']

for i in range(len(images)):
    plt.subplot(2,4,i+1)
    if len(images[i].shape) == 2:
        plt.imshow(images[i], cmap='gray')
    else:
        plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB))
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Load image
image = cv2.imread(r'D:\University\4th semester\PAI Lab\Task 4\dog.jpg')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# ------------------- 1. Line Detection (HoughLines) -------------------
edges = cv2.Canny(gray, 50, 150, apertureSize=3)
lines = cv2.HoughLines(edges, 1, np.pi/180, 120)
line_image = image.copy()
if lines is not None:
    for rho, theta in lines[:,0]:
        a = np.cos(theta)
        b = np.sin(theta)
        x0 = a*rho
        y0 = b*rho
        x1 = int(x0 + 1000*(-b))
        y1 = int(y0 + 1000*(a))
        x2 = int(x0 - 1000*(-b))
        y2 = int(y0 - 1000*(a))
        cv2.line(line_image, (x1,y1), (x2,y2), (0,0,255), 2)

# ------------------- 2. Circle Detection (HoughCircles) -------------------
blur = cv2.medianBlur(gray, 5)
circles = cv2.HoughCircles(blur, cv2.HOUGH_GRADIENT, 1, 20,
                           param1=50, param2=30, minRadius=10, maxRadius=100)
circle_image = image.copy()
if circles is not None:
    circles = np.uint16(np.around(circles))
    for i in circles[0,:]:
        cv2.circle(circle_image, (i[0], i[1]), i[2], (0,255,0), 2)
        cv2.circle(circle_image, (i[0], i[1]), 2, (0,0,255), 3)

# ------------------- 3. Corner Detection (Harris) -------------------
gray_float = np.float32(gray)
dst = cv2.cornerHarris(gray_float, 2, 3, 0.04)
dst = cv2.dilate(dst, None)
harris_image = image.copy()
harris_image[dst>0.01*dst.max()] = [0,0,255]

# ------------------- 4. Shi-Tomasi Corner Detection -------------------
shi_corners = cv2.goodFeaturesToTrack(gray, 100, 0.01, 10)
shi_image = image.copy()
if shi_corners is not None:
    for corner in shi_corners:
        x, y = corner.ravel()
        cv2.circle(shi_image, (int(x), int(y)), 3, (0,255,0), -1)

# ------------------- Display -------------------
plt.figure(figsize=(15,10))
images = [image, line_image, circle_image, harris_image, shi_image]
titles = ['Original', 'Hough Lines', 'Hough Circles', 'Harris Corners', 'Shi-Tomasi Corners']

for i in range(len(images)):
    plt.subplot(2,3,i+1)
    plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB))
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Load image
image = cv2.imread(r'D:\University\4th semester\PAI Lab\Task 4\dog.jpg')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# ------------------- 1. Find Contours -------------------
# Threshold the image for contour detection
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

contour_image = image.copy()
for i, cnt in enumerate(contours):
    cv2.drawContours(contour_image, [cnt], 0, (0,255,0), 2)
    # Print coordinates of first 5 points of each contour
    print(f"Contour {i} first 5 coordinates: {cnt[:5].reshape(-1,2)}")

# ------------------- 2. Histogram Analysis -------------------
# Grayscale histogram
hist_gray = cv2.calcHist([gray], [0], None, [256], [0,256])

# Color histogram
hist_b = cv2.calcHist([image], [0], None, [256], [0,256])
hist_g = cv2.calcHist([image], [1], None, [256], [0,256])
hist_r = cv2.calcHist([image], [2], None, [256], [0,256])

# ------------------- Display -------------------
plt.figure(figsize=(15,7))

# Contours
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(contour_image, cv2.COLOR_BGR2RGB))
plt.title('Contours')
plt.axis('off')

# Histograms
plt.subplot(1,2,2)
plt.plot(hist_gray, color='k', label='Gray')
plt.plot(hist_b, color='b', label='Blue')
plt.plot(hist_g, color='g', label='Green')
plt.plot(hist_r, color='r', label='Red')
plt.title('Histogram Analysis')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')
plt.legend()

plt.tight_layout()
plt.show()
```
