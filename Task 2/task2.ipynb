```python
def water_jug_dfs(jug1_cap, jug2_cap, target):
    stack = [((0, 0), [])]
    visited = set()

    while stack:
        (x, y), path = stack.pop()

        if x == target or y == target:
            path.append(f"Goal Reached! Final State: ({x}, {y})")
            return path

        if (x, y) in visited:
            continue
        visited.add((x, y))

        next_moves = [
            ((jug1_cap, y), f"Fill Jug 1 -> ({jug1_cap}, {y})"),
            ((x, jug2_cap), f"Fill Jug 2 -> ({x}, {jug2_cap})"),
            ((0, y), f"Empty Jug 1 -> (0, {y})"),
            ((x, 0), f"Empty Jug 2 -> ({x}, 0)"),
            ((x - min(x, jug2_cap - y), y + min(x, jug2_cap - y)),
             f"Pour Jug 1 into Jug 2 -> ({x - min(x, jug2_cap - y)}, {y + min(x, jug2_cap - y)})"),
            ((x + min(y, jug1_cap - x), y - min(y, jug1_cap - x)),
             f"Pour Jug 2 into Jug 1 -> ({x + min(y, jug1_cap - x)}, {y - min(y, jug1_cap - x)})")
        ]

        for next_state, move in next_moves:
            if next_state not in visited:
                stack.append((next_state, path + [move]))

    return None

if __name__ == "__main__":
    JUG1_CAPACITY = 4
    JUG2_CAPACITY = 3
    TARGET_AMOUNT = 2

    print("Starting DFS for the Water Jug Problem (4L, 3L) to reach 2L...\n")
    print("Initial State: (0, 0)")

    solution = water_jug_dfs(JUG1_CAPACITY, JUG2_CAPACITY, TARGET_AMOUNT)

    if solution:
        print("\n--- Solution Found ---")
        for i, step in enumerate(solution, 1):
            print(f"Step {i}: {step}")
    else:
        print("\nNo solution exists for the given jug capacities and target.")
```
